```python
from roboflow import Roboflow
import os

# Authenticate (replace with your API key)
rf = Roboflow(api_key="UcNPPzQ2W6YrPv1iD2wr")

# Load your workspace and project
project = rf.workspace("datacapstoneproject").project("soft-drinks-fridge-lcaqm")
dataset_root = "Soft-Drinks-Fridge-3-synthetic"

# === 3. Upload each split ===
splits = {
    "train": "train",
    "valid": "valid",
    "test": "test"
}

for split, folder in splits.items():
    folder_path = os.path.join(dataset_root, folder)
    if not os.path.exists(folder_path):
        print(f"⚠️ Folder not found: {folder_path}")
        continue

    print(f"🚀 Uploading {split} set from {folder_path} ...")
    project.upload(folder_path, split=split, num_threads=8)

print("✅ All done! Check your Roboflow project for uploaded images.")
```
